# Data-mart Problem

## Business Context:
The platform has 20M+ users. Business wants to label user segments on a daily basis to support better customer targeting and CRM actions.

## Business Stakeholder Example:
Example segments may include: 
- new_user
- active_buyer
- repeat_buyer
- high_value_user
- dormant_user
- churn_risk_user
- BNPL_user
- wallet_user
- discount_sensitive_user

## Expected output

Business initially requests storing one row per user per day. Please propose a scalable data mart design and explain whether this approach is suitable.

Please include: 
1. Proposed table design and grain 
2. Primary key and partition column 
3. Snapshot vs changelog vs hybrid approach 
4. Refresh / overwrite logic 
5. Data quality checks 
6. Assumptions and trade-offs
