### Create connection from the Database

```python
conn = DataBase(path=db_path).conn
```

```python
import pandas as pd
import sqlite3
def _qr_to_df(
    conn: sqlite3.Connection,
    query: str,
) -> pd.DataFrame:
    return pd.read_sql_query(
        sql=query,
        con=conn,
    )
```

```python
df = _qr_to_df(conn=conn,query="""
    select
        *
    from dim_user
""")
```
